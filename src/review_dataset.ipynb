```python
import torch
import matplotlib.pyplot as plt
import random
from amr.dataset import HybridGenomeDataset
from configs.kleb_config import kleb_config
from configs.staphy_config import staphy_config
# Assuming `image` is a 32x32 NumPy array with values in [0, 1]
def show_example_image(dataloader):
    
    # Get one batch from the dataloader
    data, labels = next(iter(dataloader))  # images: [B, C, H, W]
    images = data[0]
    # Choose a random index in the batch
    idx = random.randint(0, images.size(0) - 1)

    # Extract and convert to 2D grayscale image
    img = images[idx].squeeze()
    labels = torch.argmax(labels, dim=1)
    print(img.shape)
    print(img.max())
    print(img.min())
    plt.imshow(img, cmap='gray', vmin=0, vmax=255)
    plt.axis('off')
    plt.title(f"Label: {labels[idx].item()}")
    plt.show()
    
    
staphy_dataset = HybridGenomeDataset(
    root_dir=staphy_config["root_dir"],
    train_or_test="train",
    pathogen=staphy_config["pathogen"],
    genes=staphy_config["genes"],
    k=staphy_config["k"]
)

kleb_dataset = HybridGenomeDataset(
    root_dir=kleb_config["root_dir"],
    train_or_test="train",
    pathogen=kleb_config["pathogen"],
    genes=kleb_config["genes"],
    k=kleb_config["k"]
)

staphy_loader = get_dataloader(staphy_dataset,batch_size=64)
kleb_loader = get_dataloader(kleb_dataset,batch_size=64)
show_example_image(staphy_loader)
show_example_image(kleb_loader)
```
